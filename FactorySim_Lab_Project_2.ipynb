```python

# FactorySim Lab - Project 2
# Purpose: Simulate a factory machine producing items.

machine_name = "Assembly Machine 1"
machine_status = "Running"
products_produced = 0

print("Welcome to FactorySim Lab!")
print("--------------------------")
print("Machine:", machine_name)
print("Status:", machine_status)

# Simulate five production cycles
for cycle in range(5):
    products_produced = products_produced + 1
    print("Production cycle", cycle + 1, "- Item produced!")

print("--------------------------")
print("Total products produced:", products_produced)
```
